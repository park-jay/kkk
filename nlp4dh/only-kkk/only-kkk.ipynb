```python
import pandas as pd
```

```python
data=pd.read_feather('/Volumes/T7/chroniclingamerica/american-stories/kkk-revival-stanza-lynch.feather')
```

```python
pattern=r'\b(klan|klux|ku)\b'
only_kkk=data[data['article'].str.contains(pattern, regex=True, case=False)]
```

```python
only_kkk.reset_index(drop=True, inplace=True)
```

```python
only_kkk.to_feather('/Volumes/T7/chroniclingamerica/american-stories/only-kkk-revival.feather')
```
